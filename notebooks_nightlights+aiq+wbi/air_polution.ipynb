### Merge the datasets

```python
df_ap_europe = pd.concat(
    [
        df_ap_e_2017_1,
        df_ap_e_2018_1,
        df_ap_e_2019_1,
        df_ap_e_2019_2,
        df_ap_e_2019_3,
        df_ap_e_2019_4,
        df_ap_e_2020_1,
        df_ap_e_2020_2,
        df_ap_e_2020_3,
        df_ap_e_2020_4,
        df_ap_e_2021_1,
        df_ap_e_2021_2,
        df_ap_e_2021_3,
        df_ap_e_2021_4,
        #df_ap_e_2022_1,
        df_ap_e_2022_2,
        df_ap_e_2022_3,
        df_ap_e_2022_4
    ],
    ignore_index=True
)

df_ap_europe
```

```python
df_ap_europe["Date"] = pd.to_datetime(df_ap_europe["Date"])
df_ap_europe["Year"] = df_ap_europe["Date"].dt.year
```

```python
df_year = (
    df_ap_europe
    .groupby(["Year", "Country", "City", "Specie"], as_index=False)
    .agg({
        "count": "mean",
        "min": "mean",
        "max": "mean",
        "median": "mean",
        "variance": "mean"
    })
)

df_year
```

### Pollutants

```python
pollutants = ["pm10", "pm25", "no2", "co", "so2", "o3"]

df_pollution = df_year[df_year["Specie"].isin(pollutants)].copy()
```

#### Median

```python
df_year_median = (
    df_pollution
    .groupby(["Year", "Country", "City", "Specie"], as_index=False)
    ["median"]
    .mean()
)

df_final = df_year_median.pivot(
    index=["Country", "City", "Year"],
    columns="Specie",
    values="median"
).reset_index()

df_final
```

```python
df_final.isna().sum()
```

#### pm10

```python
df_clean_pm10 = df_final.dropna(subset=["pm10"]).copy()
```
