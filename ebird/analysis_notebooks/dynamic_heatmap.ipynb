```python
import pandas as pd
import numpy as np
import folium
from folium.plugins import HeatMapWithTime
```

```python

df_native = pd.read_csv(r"C:\Users\user\Downloads\native_species_europe_only_cleaned.csv")
df_migratory = pd.read_csv(r"C:\Users\user\Downloads\migra_species_europe_only_cleaned.csv")
```

```python



def prepare_heatmap_data(df):
    
    df['year_month'] = df['year'].astype(str) + '-' + df['month'].astype(str).str.zfill(2)
    
    # Sort by date to ensure the slider moves forward in time
    df = df.sort_values('year_month')
    
    time_index = sorted(df['year_month'].unique())
    data_list = []

    for period in time_index:
        # Filter for the month/year
        subset = df[df['year_month'] == period]
        
       
        lat_lon_weight = subset[['LATITUDE', 'LONGITUDE', 'howMany']].fillna(1).values.tolist()
        data_list.append(lat_lon_weight)
        
    return data_list, time_index

# Process the migratory birds
migratory_data, time_labels = prepare_heatmap_data(df_migratory)
```

```python
# Initialize Map
m = folium.Map(location=[54.5, 15.2], zoom_start=4, tiles="cartodbpositron")

# Add Migratory Heatmap
hm_migratory = HeatMapWithTime(
    data=migratory_data,
    index=time_labels,
    radius=12,
    auto_play=False,
    max_opacity=0.7,
    name="Migratory Birds"
).add_to(m)



folium.LayerControl().add_to(m)
m.save("bird_movement_2018_2022.html")
```
