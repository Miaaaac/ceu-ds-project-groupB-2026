## Multivariate regression predicting how far a bird observation occurs from the nearest city based on pollution and environmental conditions

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Dependent variable
y = df["distance_km"].fillna(df["distance_km"].mean())

# Independent variables
features = ["pollution_score_weighted", "co", "no2", "so2", "pm10", "pm25", "o3"]
optional_features = ["temperature", "humidity", "pressure"]
features += [f for f in optional_features if f in df.columns]

X = df[features].fillna(df[features].mean())

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Fit regression
model = LinearRegression()
model.fit(X_scaled, y)

# Coefficients
coef_df = pd.DataFrame({"Feature": features, "Coefficient": model.coef_})
print(coef_df)

# Evaluate
y_pred = model.predict(X_scaled)
print("R²:", r2_score(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# ----------------------------
# 1. Select target
# ----------------------------
y = np.log1p(df["distance_km"].fillna(0))  # log(1 + distance_km)

# ----------------------------
# 2. Pollution variables (PCA)
# ----------------------------
pollution_vars = ["pollution_score_weighted", "co", "no2", "so2", "pm10", "pm25", "o3"]
pollution_vars = [v for v in pollution_vars if v in df.columns]

# Standardize
scaler_pollution = StandardScaler()
pollution_scaled = scaler_pollution.fit_transform(df[pollution_vars])

# PCA
pca = PCA(n_components=1)  # 1 component captures main variation
pollution_pca = pca.fit_transform(pollution_scaled)
df_pollution_pca = pd.DataFrame(pollution_pca, columns=["pollution_PCA"])

# ----------------------------
# 3. Other numeric features
# ----------------------------
numeric_vars = ["temperature", "humidity", "pressure", "ntl_intensity", "month"]
numeric_vars = [v for v in numeric_vars if v in df.columns]

df_numeric = df[numeric_vars].fillna(df[numeric_vars].mean())

# Standardize
scaler_numeric = StandardScaler()
numeric_scaled = scaler_numeric.fit_transform(df_numeric)
df_numeric_scaled = pd.DataFrame(numeric_scaled, columns=numeric_vars)

# ----------------------------
# 4. Combine features
# ----------------------------
X = pd.concat([df_pollution_pca, df_numeric_scaled], axis=1)

# Optionally include categorical info like species_type (one-hot)
if "species_type" in df.columns:
    df_cat = pd.get_dummies(df["species_type"], drop_first=True)
    X = pd.concat([X, df_cat.reset_index(drop=True)], axis=1)

# ----------------------------
# 5. Fit regression
# ----------------------------
model = LinearRegression()
model.fit(X, y)

# ----------------------------
# 6. Coefficients
# ----------------------------
coef_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})
print(coef_df)

# ----------------------------
# 7. Model evaluation
# ----------------------------
y_pred = model.predict(X)
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)
print("R²:", r2)
print("MSE:", mse)
```

small coefficients indicate that pollution and these environmental variables explain very little of the variance in distance.

R² = 0.0196 → ~2% of variance explained

Still low, but expected in ecological datasets with many unobserved factors (habitat, sampling design, topography).

MSE ≈ 0.95 → since we log-transformed distance, this is the mean squared log-error.

## Second reg

Pollution metrics: pollution_score_mean, pollution_score_weighted, or pollution_score_pca (avoid multicollinearity)

Nighttime lights intensity: ntl_intensity

Species characteristics: species_type, OBSERVATION TYPE

Temporal effects: Year, month

Optional: Remote flag: is_remote

We’ll log-transform distance_km, standardize numeric features, one-hot encode categorical variables, and fit a regression with optional Ridge regularization to stabilize coefficients.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# ----------------------------
# 1. Target
# ----------------------------
y = np.log1p(df["distance_km"].fillna(0))  # log(1 + distance_km)

# ----------------------------
# 2. Features (numeric and interpretable)
# ----------------------------
features = [
    'ntl_intensity',                 # main focus
    'pollution_score_mean',          # city-level average pollution
    'pollution_score_weighted',      # weighted pollution
    'pollution_score_pca',           # optional: PCA of pollutants
    'co', 'no2', 'so2', 'pm10', 'pm25', 'o3',  # specific pollutants
    'Year',
    'month',
    'temperature', 'humidity', 'pressure'      # weather, if available
]

# Keep only columns that exist in df
features = [f for f in features if f in df.columns]

X = df[features].fillna(df[features].mean())

# ----------------------------
# 3. Standardize features
# ----------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ----------------------------
# 4. Fit multivariate regression
# ----------------------------
model = LinearRegression()
model.fit(X_scaled, y)

# ----------------------------
# 5. Coefficients and insights
# ----------------------------
coef_df = pd.DataFrame({
    'Feature': features,
    'Coefficient': model.coef_
}).sort_values(by='Coefficient', key=abs, ascending=False)

print("Feature coefficients:")
print(coef_df)

# ----------------------------
# 6. Model evaluation
# ----------------------------
y_pred = model.predict(X_scaled)
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)

print("\nModel Performance:")
print("R²:", r2)
print("MSE:", mse)

# ----------------------------
# 7. Correlations with target
# ----------------------------
corr_with_target = df[features + ['distance_km']].corr()['distance_km'].sort_values(key=abs, ascending=False)
print("\nCorrelation with distance_km:")
print(corr_with_target)
```

Most effects are small. The largest in magnitude is NO2, suggesting urban traffic pollution may slightly influence bird distance. Nightlights (ntl_intensity) has a very small positive coefficient here.

Model Performance

R² ≈ 0.038 → ~3.8% of variance in distance explained.

MSE ≈ 0.933 → average squared log-distance error.

This is low, which is expected: bird distance is influenced by many other ecological, geographic, and species-specific factors, not just pollution and nightlights.

Strongest positive raw correlations: ozone and nightlights.

Strongest negative raw correlation: SO2 and NO2.

Some coefficients in the multivariate model differ from correlations due to controlling for other variables (multicollinearity).

Nightlights effect is very subtle in this dataset; urban light alone doesn’t strongly predict distance.

Pollution has some influence, but effects vary across pollutants (NO2 vs PM10).

Year and month have minimal effect, small temporal trend.

Overall, this tells us: bird observation distance is influenced more by other factors, such as species behavior, habitat, and city geography.

```python
import statsmodels.api as sm
import numpy as np
import pandas as pd

# ----------------------------
# 1. Features and target
# ----------------------------
features = [
    'ntl_intensity', 'pollution_score_mean', 'pollution_score_weighted',
    'pollution_score_pca', 'co', 'no2', 'so2', 'pm10', 'pm25', 'o3',
    'Year', 'month', 'temperature', 'humidity', 'pressure'
]
features = [f for f in features if f in df.columns]

X = df[features].fillna(df[features].mean())
y = np.log1p(df['distance_km'].fillna(0))

# ----------------------------
# 2. Add constant for intercept
# ----------------------------
X = sm.add_constant(X)

# ----------------------------
# 3. Fit OLS regression with robust standard errors (HC3)
# ----------------------------
model = sm.OLS(y, X).fit(cov_type='HC3')  # HC3 for heteroscedasticity-robust

# ----------------------------
# 4. Summary
# ----------------------------
print(model.summary())
```

Dependent variable: distance_km (log-transformed in scikit-learn earlier, raw here).

Observations: 1,656,552 → huge sample, so tiny effects are often statistically significant.

R² ≈ 0.038 → your numeric features explain ~3.8% of variance in distance. This is small but expected; ecological data are noisy.

Covariance Type: HC3 → heteroscedasticity-robust standard errors. This makes your p-values more reliable.

Warning: small eigenvalue → multicollinearity among pollution variables (e.g., PM10, PM2.5, NO2, CO) — something to watch.

Feature	Coef	z-value	p-value	Interpretation
ntl_intensity	0.0002	4.505	0.000	Statistically significant; small positive effect: brighter cities → slightly farther bird observations.
pollution_score_mean	0.0014	2.591	0.010	Weak but significant: more pollution slightly increases distance.
pollution_score_weighted	0.0251	54.265	0.000	Stronger effect; weighted pollution increases distance.
pollution_score_pca	0.0412	52.365	0.000	Significant: overall pollution PCA increases distance.

```python
import matplotlib.pyplot as plt

# Extract coefficients and 95% CI
coef = model.params
conf = model.conf_int()
features = coef.index

plt.figure(figsize=(10,6))
plt.barh(features, coef, xerr=[coef - conf[0], conf[1] - coef], color='skyblue')
plt.axvline(0, color='gray', linestyle='--')
plt.xlabel('Coefficient')
plt.title('OLS Coefficients with 95% Confidence Intervals')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# ----------------------------
# 1. Define numeric threshold using IQR
# ----------------------------
Q1 = df['distance_km'].quantile(0.25)
Q3 = df['distance_km'].quantile(0.75)
IQR = Q3 - Q1

# Keep points within 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_clean = df[(df['distance_km'] >= lower_bound) & (df['distance_km'] <= upper_bound)]
print(f"Dropped {len(df) - len(df_clean)} outliers. New dataset size: {len(df_clean)}")
```

```python
import statsmodels.api as sm

# Features
features = [
    'ntl_intensity', 'pollution_score_mean', 'pollution_score_weighted',
    'pollution_score_pca',
    'Year', 'month', 'temperature', 'humidity', 'pressure'
]
features = [f for f in features if f in df.columns]

X = df_clean[features].fillna(df_clean[features].mean())
y = np.log1p(df_clean['distance_km'].fillna(0))  # log transform optional

# Add constant
X = sm.add_constant(X)

# Fit OLS with robust errors
model_clean = sm.OLS(y, X).fit(cov_type='HC3')

# Summary
print(model_clean.summary())
```

```python
df_clean = df_clean[df_clean['distance_km'] < df_clean['distance_km'].quantile(0.99)]
df_clean['ntl_intensity_log'] = np.log1p(df_clean['ntl_intensity'])
df_clean['pollution_score_weighted_log'] = np.log1p(df_clean['pollution_score_weighted'])
features = ['ntl_intensity_log', 'pollution_score_weighted_log', 'Year', 'month']
X = df_clean[features].fillna(df_clean[features].mean())
y = np.log1p(df_clean['distance_km'])

X = sm.add_constant(X)
model_final = sm.OLS(y, X).fit(cov_type='HC3')
print(model_final.summary())
```

Observations: 1,557,158

R² = 0.024 → explains ~2.4% of the variation in log-distance.

Condition number is still high (3.19e6) → some multicollinearity, but simpler model reduces issues.

Nightlights matter: log-transformed nightlights has the largest magnitude coefficient. Urban areas (higher NTL intensity) are associated with closer bird observations, which matches intuition.

Pollution effect is weaker, but still significant.

Temporal effects (Year, month) are small but detectable.

Log-transform helps reduce influence of extreme distances.
